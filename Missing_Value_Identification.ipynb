# Task Overview

This project focuses on identifying and summarizing missing values in the Titanic dataset using Python and Pandas. It demonstrates basic missing-data inspection and explains why missing values should be investigated before deciding how to handle them.

#### Objectives

Identify missing values in each column.

Calculate missing-value counts and percentages.

Sort columns by the number of missing values.

Summarize findings and understand the risks of blindly deleting data.

## Importing Libraries

```python
import pandas as pd
import numpy as np
```

## Loading Dataset

```python
df = pd.read_csv("titanic.csv")
df.head()
```

## Dataset Information

```python
print("Dataset Shape :",df.shape)
df.info()
```

## Missing Values Count by Columns

```python
missing_count = df.isnull().sum()
print(missing_count)
```

## Missing Value Summary

```python
missing_summary = pd.DataFrame({
    "Missing Values":df.isnull().sum(),
    "Missing Percentage":(df.isnull().sum()/len(df)*100).round(2)
})
missing_summary = missing_summary.sort_values(by="Missing Values",ascending = False)
missing_summary
```

## Total Missing Values

```python
print("Total Missing Values :",df.isnull().sum().sum())
```

## Columns with Missing Values

```python
print("Columns with Missing Values",(df.isnull().sum()>0).sum())
```

## Findings — Missing Value Identification

The Titanic dataset contains 891 rows and 15 columns. Missing-value analysis identified 869 missing cells across four columns: `deck`, `age`, `embarked`, and `embark_town`.

The `deck` column has the highest number of missing values, with 688 missing entries (77.22%). The `age` column contains 177 missing values (19.87%). Both `embarked` and `embark_town` contain two missing values each (0.22%).

The remaining 11 columns contain no missing values. The results show that missing data is concentrated mainly in the `deck` and `age` columns. Before removing or imputing any values, the reasons for missingness and the importance of each feature should be considered.

No values were removed or modified during this inspection.

## Interview Questions and Answers

1. What is a missing value?

A missing value is a data entry that is absent or unavailable in a dataset. In Python, Pandas commonly represents missing values using NaN or None.

Example: If a passenger's age is not recorded, the Age column contains a missing value.

2. Why can blindly deleting rows with missing values be risky?

Blindly deleting rows with missing values can result in the loss of useful information, reduce the dataset size, and introduce bias into the analysis. Therefore, missing values should be investigated before deciding whether to remove rows or use an appropriate imputation method.

Example: In the Titanic dataset, removing every row with a missing age could discard useful passenger information, including survival status.
